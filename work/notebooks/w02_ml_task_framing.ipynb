## 5. Why ML beats a fixed rule here

A fixed rule can be useful as a transparent baseline, but it may treat every page using the same manually chosen thresholds and weights.

ML can combine multiple observable signals at the same time and learn patterns from the available examples. For example, impressions, sessions, position, CTR, content age, and engagement signals may interact in ways that are difficult to capture with one simple rule.

The purpose is not to use ML just because it is more complicated. ML is useful if it produces a better-ranked review queue on unseen data than a transparent baseline. If it does not improve the relevant validation metric, the fixed rule may be the better choice.
