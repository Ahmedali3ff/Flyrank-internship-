## 2. The question: decision, action, cost of a wrong call

Research question: Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring?

Decision: Decide which pages deserve priority for human review.

Action: A content team can review the highest-ranked pages and choose an appropriate action such as refreshing, expanding, protecting, pruning, or monitoring the page.

Cost of a wrong recommendation: A false positive can waste limited review and content resources on a low-value page. A false negative can cause an important page or opportunity to be missed. The goal is therefore to improve prioritization, not to guarantee that an action will improve performance.
