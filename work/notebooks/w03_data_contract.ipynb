```python
import pandas as pd

DATA_PATH = "data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())
```

## 1. Unit of analysis + time window

**Unit of analysis:** One row represents one anonymized content observation associated with one client.

**Time window:** The dataset does not contain an explicit calendar date column. Instead, it contains rolling performance windows, including 90-day and recent/previous 30-day metrics, plus content age and days since last update. Therefore, the analysis treats each row as a historical content observation summarized over these available windows rather than assigning an assumed calendar date range.

```python
# Verify unit of analysis and available observation windows

print("Rows:", len(df))
print("Unique content_id:", df["content_id"].nunique())
print("Unique client_id:", df["client_id"].nunique())

print("\nWindow-related fields:")
window_fields = [
    c for c in df.columns
        if any(term in c.lower() for term in ["30d", "90d", "days_"])
        ]

print(window_fields)

print("\nContent age:")
print(df["content_age_days"].describe())

print("\nDays since last update:")
print(df["days_since_last_update"].describe())
```

## 2. Fields: feature / label / context / excluded

### Label

- `trend_direction` — used to reconstruct the binary declining/not-declining outcome.

### Context

- `content_id` — anonymized content identifier used for traceability/grouping, not as a predictive feature.
- `client_id` — anonymized client identifier used for grouped validation, not as a predictive feature.

### Excluded

- `trend_direction` — excluded from the feature set because it directly defines the target.
- `trend_pct` — excluded because it directly represents the measured trend outcome and would create target leakage.

### Features

All remaining fields are treated as candidate predictive features, subject to the leakage checks and modeling constraints documented in the later notebooks.

```python
# Define the data contract buckets

label_field = "trend_direction"

context_fields = [
    "content_id",
        "client_id"
        ]

excluded_fields = [
            "trend_direction",
                "trend_pct"
                ]

feature_fields = [
                    c for c in df.columns
                        if c not in context_fields + excluded_fields
                        ]

print("LABEL:")
print([label_field])

print("\nCONTEXT:")
print(context_fields)

print("\nEXCLUDED:")
print(excluded_fields)

print("\nFEATURE COUNT:", len(feature_fields))
print("FEATURES:")
print(feature_fields)
```

## 4. Data limits

This dataset supports historical, directional analysis and decision support, but it cannot establish causal relationships.

Important limitations include:

- The dataset does not provide an explicit calendar-date field, so the analysis cannot claim a specific calendar release period from the available schema.
- Performance variables are summarized over rolling 90-day and 30-day windows rather than a single synchronized event date.
- Recent and previous 30-day windows may overlap with the broader 90-day window, so observations should not automatically be interpreted as independent experiments.
- Content age and days since last update provide historical context but do not establish why performance changed.
- The declining label represents an observed historical pattern and does not identify a causal mechanism.
- The data cannot establish that a particular content change, ranking-system change, Google update, or external event caused the observed outcome.
- Some performance fields may be unavailable or sparse for portions of the content history, so missingness and coverage should be considered when interpreting results.
- Model predictions should be used as review-prioritization and decision-support signals, not autonomous decisions.
